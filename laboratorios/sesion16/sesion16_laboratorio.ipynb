## Fase 1 · El Artesano (15 min, sin IA)

Escribe a mano la implementación en `artesano_sesion16_c.c`. El archivo ya trae la

````python
%%writefile api_sesion16.h
/* api_sesion16.h - contrato de la Sesion 16
 * Registro de prototipos y aridad
 *
 * Esta cabecera es el contrato: el harness y tu implementacion
 * deben incluirla. Si cambias una firma, el harness dejara de compilar.
 */
#ifndef API_SESION16_H
#define API_SESION16_H

typedef struct Funcion {
    const char *nombre;
    int aridad;
} Funcion;

int verificar_llamada(const Funcion *catalogo, int total, const char *nombre, int nargs);

#endif /* API_SESION16_H */
````

````python
%%writefile artesano_sesion16_c.c

/* artesano_sesion16_c.c - Fase 1: El Artesano */

#include "api_sesion16.h"
#include <string.h>

int verificar_llamada(const Funcion *catalogo, int total,
                      const char *nombre, int nargs) {

    int i;

    /* Validar los datos recibidos */
    if (catalogo == NULL || nombre == NULL || nargs < 0) {
        return -1;
    }

    /* Recorrer el catalogo de funciones */
    for (i = 0; i < total; i++) {

        /* Comparar el nombre de la funcion */
        if (strcmp(catalogo[i].nombre, nombre) == 0) {

            /* Comparar la cantidad de argumentos */
            if (catalogo[i].aridad == nargs) {
                return 1;
            } else {
                return 0;
            }
        }
    }

    /* La funcion no se encontro */
    return -1;
}

````

## Fase 2 · El Científico (10 min)

Redacta el contrato **C-R-E-O** (Contexto, Rol, Especificación, Output) para pedirle

**CONTRATO C-R-E-O**

**Contexto:** Estoy desarrollando la Sesión 16 de Lenguajes y Autómatas II. Necesito verificar las llamadas a funciones mediante un catálogo que almacena el nombre de cada función y su número de parámetros.

La estructura proporcionada es:
typedef struct Funcion {
    const char *nombre;
    int aridad;
} Funcion;

**Rol: ** Actúa como un experto en programación C, estructuras, cadenas de caracteres y validación de parámetros formales.

**Especificación:** Implementa la siguiente función respetando exactamente su firma:
int verificar_llamada(const Funcion *catalogo, int total, const char *nombre, int nargs);
La función debe cumplir estas condiciones:
1. Devolver -1 si el catálogo es NULL.
2. Devolver -1 si el nombre es NULL.
3. Devolver -1 si la cantidad de argumentos es negativa.
4. Recorrer el catálogo y buscar el nombre exacto de la función.
5. Comparar los nombres utilizando strcmp().
6. Devolver 1 cuando la función exista y su aridad coincida con nargs.
7. Devolver 0 cuando la función exista, pero su aridad sea diferente.
8. Devolver -1 cuando el nombre no esté registrado.
9. Aceptar correctamente funciones con cero parámetros.
10. Validar también que el total de elementos no sea negativo y evitar comparar nombres almacenados que sean NULL.
No cambies la estructura ni la firma original. No agregues main(), no utilices bibliotecas no estándar y evita advertencias al compilar con GCC usando -std=c11 -Wall -Wextra -Werror.

**Output:** Entrega el código completo para ia_sesion16_c.c, con comentarios explicativos y un análisis de complejidad temporal y espacial.

````python
%%writefile ia_sesion16_c.c

/* ia_sesion16_c.c - Fase 2: El Cientifico */

#include "api_sesion16.h"
#include <string.h>

int verificar_llamada(const Funcion *catalogo, int total,
                      const char *nombre, int nargs) {

    int i;

    /* Validar los argumentos recibidos */
    if (catalogo == NULL || nombre == NULL ||
        total < 0 || nargs < 0) {
        return -1;
    }

    /* Buscar la funcion en el catalogo */
    for (i = 0; i < total; i++) {

        /* Evitar comparar nombres nulos */
        if (catalogo[i].nombre == NULL) {
            continue;
        }

        /* Comparar el nombre exacto */
        if (strcmp(catalogo[i].nombre, nombre) == 0) {

            /* Verificar que coincida la aridad */
            if (catalogo[i].aridad == nargs) {
                return 1;
            }

            return 0;
        }
    }

    /* La funcion no esta registrada */
    return -1;
}

````

## Fase 3 · El Auditor (10 min)

El harness de abajo **no aprueba nada por sí solo**: ejecuta tu función caso por caso,
la compara contra el valor esperado e imprime `OK` o `FALLA`. Léelo antes de ejecutarlo:
los casos te dicen qué se está midiendo.


````python
%%writefile harness_sesion16_c.c
/* harness_sesion16_c.c - auditoria automatica de la Sesion 16
 *
 * Compilacion:
 *   gcc -std=c11 -Wall -Wextra -Werror harness_sesion16_c.c artesano_sesion16_c.c -o test_artesano
 *
 * El harness NO aprueba nada por si solo: cada linea se calcula
 * ejecutando tu implementacion y comparandola contra el valor esperado.
 * Si algun caso falla, el programa termina con codigo de salida 1.
 */
#include <stdio.h>
#include <string.h>
#include "api_sesion16.h"

static int probadas = 0;
static int fallidas = 0;

static void verificar_int(const char *caso, long esperado, long obtenido) {
    int ok = (esperado == obtenido);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %-34s esperado: %-12ld obtenido: %-12ld %s\n",
           probadas, caso, esperado, obtenido, ok ? "OK" : "FALLA");
}

int main(void) {
    printf("=== AUDITORIA SESION 16 - Registro de prototipos y aridad ===\n");
    printf("Caso                               esperado:              obtenido:              resultado\n");

    { Funcion c[] = {{"suma", 3}}; int r = verificar_llamada(c, 1, "suma", 3); verificar_int("suma con 3 argumentos", (long)(1), (long)(r)); }
    { Funcion c[] = {{"suma", 3}}; int r = verificar_llamada(c, 1, "suma", 2); verificar_int("suma con 2 argumentos", (long)(0), (long)(r)); }
    { Funcion c[] = {{"suma", 3}}; int r = verificar_llamada(c, 1, "suma", 4); verificar_int("suma con 4 argumentos", (long)(0), (long)(r)); }
    { Funcion c[] = {{"suma", 3}}; int r = verificar_llamada(c, 1, "resta", 2); verificar_int("funcion no registrada", (long)(-1), (long)(r)); }
    { Funcion c[] = {{"main", 0}}; int r = verificar_llamada(c, 1, "main", 0); verificar_int("main sin argumentos", (long)(1), (long)(r)); }
    { Funcion c[] = {{"suma", 3}}; int r = verificar_llamada(c, 1, NULL, 3); verificar_int("nombre NULL", (long)(-1), (long)(r)); }

    printf("\nRESULTADO: %d/%d pruebas superadas\n",
           probadas - fallidas, probadas);
    if (fallidas != 0) {
        printf("Aun hay %d caso(s) sin resolver. Revisa tu implementacion.\n",
               fallidas);
        return 1;
    }
    printf("TODO EN VERDE. Documenta tu auditoria en auditorias/auditoria_sesion16.md" "\n");
    return 0;
}
````
