## 2. El contrato exacto

El archivo `api_sesion23.h` declara la interfaz que el harness sabe probar.


## Fase 1 · El Artesano (15 min, sin IA)

Escribe a mano la implementación en `artesano_sesion23_c.c`. El archivo ya trae la

````python
%%writefile api_sesion23.h
/* api_sesion23.h - contrato de la Sesion 23
 * Etiquetas y saltos incondicionales
 *
 * Esta cabecera es el contrato: el harness y tu implementacion
 * deben incluirla. Si cambias una firma, el harness dejara de compilar.
 */
#ifndef API_SESION23_H
#define API_SESION23_H

void nueva_etiqueta(char *dst, int cap, int *contador);
void emitir_flujo(char *dst, int cap, const char *etiqueta, int es_salto);

#endif /* API_SESION23_H */
````

````python
%%writefile artesano_sesion23_c.c

/* artesano_sesion23_c.c
 * Fase 1: El Artesano
 * Etiquetas y saltos incondicionales.
 */

#include "api_sesion23.h"
#include <stdio.h>

void nueva_etiqueta(char *dst, int cap, int *contador) {

    if (dst == NULL || cap <= 0 || contador == NULL) {
        return;
    }

    if (*contador < 0) {
        dst[0] = '\0';
        return;
    }

    int resultado = snprintf(dst, (size_t)cap, "L%d", *contador);

    if (resultado < 0 || resultado >= cap) {
        dst[0] = '\0';
        return;
    }

    (*contador)++;
}

void emitir_flujo(char *dst, int cap,
                  const char *etiqueta, int es_salto) {

    if (dst == NULL || cap <= 0) {
        return;
    }

    if (etiqueta == NULL) {
        dst[0] = '\0';
        return;
    }

    if (es_salto != 0) {
        snprintf(dst, (size_t)cap, "GOTO %s", etiqueta);
    } else {
        snprintf(dst, (size_t)cap, "%s:", etiqueta);
    }
}
````

## Fase 2 · El Científico (10 min)

Redacta el contrato **C-R-E-O** (Contexto, Rol, Especificación, Output) para pedirle

CONTRATO C-R-E-O

**Contexto (C):** Estoy trabajando en la Sesión 23 de Lenguajes y Autómatas II. Necesito implementar dos funciones en C11 utilizando GCC y bibliotecas estándar.

**Rol (R):** Asume el papel de un ingeniero de software experto en generación de código intermedio y control de flujo.

**Especificación (E):** Implementa las funciones `nueva_etiqueta()` y `emitir_flujo()` respetando exactamente las siguientes firmas:

```c
void nueva_etiqueta(char *dst, int cap, int *contador);
void emitir_flujo(char *dst, int cap, const char *etiqueta, int es_salto);
```

La función `nueva_etiqueta()` debe generar etiquetas consecutivas L0, L1, L2, etc., utilizando el contador recibido e incrementándolo después de generar correctamente la etiqueta.

La función `emitir_flujo()` debe generar `GOTO Lx` cuando `es_salto` sea diferente de cero y `Lx:` cuando sea cero.

Si la etiqueta es NULL, debe producir una cadena vacía. También debe validar los punteros y la capacidad del búfer para evitar accesos inválidos y desbordamientos.

No modifiques las firmas, no incluyas `main()` y utiliza `snprintf()` para controlar la escritura en memoria.

**Output (O):** Entrega el código completo y modular para `ia_sesion23_c.c`, una explicación de su funcionamiento, las mejoras realizadas respecto a una implementación manual y el análisis de complejidad temporal y espacial.


````python
%%writefile ia_sesion23_c.c

/* ia_sesion23_c.c
 * Fase 2: El Cientifico
 * Etiquetas y saltos incondicionales.
 */

#include "api_sesion23.h"
#include <stdio.h>
#include <limits.h>

void nueva_etiqueta(char *dst, int cap, int *contador) {

    int resultado;

    /* Validar los parametros recibidos */
    if (dst == NULL || cap <= 0 || contador == NULL) {
        return;
    }

    /* Inicializar el destino */
    dst[0] = '\0';

    /* Evitar contadores negativos o desbordamiento */
    if (*contador < 0 || *contador == INT_MAX) {
        return;
    }

    /* Generar la etiqueta */
    resultado = snprintf(dst, (size_t)cap, "L%d", *contador);

    /* Comprobar si la escritura fue correcta */
    if (resultado < 0 || resultado >= cap) {
        dst[0] = '\0';
        return;
    }

    /* Incrementar solo si la etiqueta se genero correctamente */
    (*contador)++;
}

void emitir_flujo(char *dst, int cap,
                  const char *etiqueta, int es_salto) {

    int resultado;

    /* Validar el destino y su capacidad */
    if (dst == NULL || cap <= 0) {
        return;
    }

    /* Inicializar el destino */
    dst[0] = '\0';

    /* Si no existe etiqueta, dejar cadena vacia */
    if (etiqueta == NULL) {
        return;
    }

    /* Generar el salto o la marca de destino */
    if (es_salto != 0) {
        resultado = snprintf(dst, (size_t)cap,
                             "GOTO %s", etiqueta);
    } else {
        resultado = snprintf(dst, (size_t)cap,
                             "%s:", etiqueta);
    }

    /* Evitar conservar resultados truncados */
    if (resultado < 0 || resultado >= cap) {
        dst[0] = '\0';
    }
}

````

## Fase 3 · El Auditor (10 min)

El harness de abajo **no aprueba nada por sí solo**: ejecuta tu función caso por caso,
la compara contra el valor esperado e imprime `OK` o `FALLA`. Léelo antes de ejecutarlo:
los casos te dicen qué se está midiendo.


````python
%%writefile harness_sesion23_c.c
/* harness_sesion23_c.c - auditoria automatica de la Sesion 23
 *
 * Compilacion:
 *   gcc -std=c11 -Wall -Wextra -Werror harness_sesion23_c.c artesano_sesion23_c.c -o test_artesano
 *
 * El harness NO aprueba nada por si solo: cada linea se calcula
 * ejecutando tu implementacion y comparandola contra el valor esperado.
 * Si algun caso falla, el programa termina con codigo de salida 1.
 */
#include <stdio.h>
#include <string.h>
#include "api_sesion23.h"

static int probadas = 0;
static int fallidas = 0;

static void verificar_str(const char *caso, const char *esperado,
                          const char *obtenido) {
    int ok = (obtenido != NULL) && (strcmp(esperado, obtenido) == 0);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %-34s esperado: %-12s obtenido: %-12s %s\n",
           probadas, caso, esperado, obtenido ? obtenido : "(NULL)",
           ok ? "OK" : "FALLA");
}

int main(void) {
    printf("=== AUDITORIA SESION 23 - Etiquetas y saltos incondicionales ===\n");
    printf("Caso                               esperado:              obtenido:              resultado\n");

    { int c = 0; char b[32] = ""; nueva_etiqueta(b, 32, &c); verificar_str("nueva_etiqueta(L0)", "L0", b); }
    { int c = 5; char b[32] = ""; nueva_etiqueta(b, 32, &c); verificar_str("nueva_etiqueta(L5)", "L5", b); }
    { char b[32] = ""; emitir_flujo(b, 32, "L2", 1); verificar_str("emitir_flujo(GOTO L2)", "GOTO L2", b); }
    { char b[32] = ""; emitir_flujo(b, 32, "L2", 0); verificar_str("emitir_flujo(marca L2)", "L2:", b); }
    { char b[32] = ""; emitir_flujo(b, 32, NULL, 1); verificar_str("emitir_flujo(NULL)", "", b); }
    { int c = 10; char b[32] = ""; nueva_etiqueta(b, 32, &c); verificar_str("nueva_etiqueta(L10)", "L10", b); }

    printf("\nRESULTADO: %d/%d pruebas superadas\n",
           probadas - fallidas, probadas);
    if (fallidas != 0) {
        printf("Aun hay %d caso(s) sin resolver. Revisa tu implementacion.\n",
               fallidas);
        return 1;
    }
    printf("TODO EN VERDE. Documenta tu auditoria en auditorias/auditoria_sesion23.md" "\n");
    return 0;
}
````
