## 1. ¿Qué vas a construir?

Para un arreglo 1D emites la multiplicación del índice por el ancho y la lectura base[_tK]. Para uno 2D primero aplanas i * filas + j, luego multiplicas por el ancho y lees. Los temporales se numeran desde *temporales.

Al final de la sesión tendrás: **Dos funciones en C que emiten el cálculo de offset y la lectura indexada para arreglos 1D y 2D.**

int emitir_acceso_2d(char lineas[][64], int max, const char *base, const char *i, const char *j, int filas, int ancho, int *temporales);

- Multiplica siempre el índice por el ancho del elemento antes de usarlo como offset.

````python
%%writefile api_sesion27.h
/* api_sesion27.h - contrato de la Sesion 27
 * Desplazamientos para arreglos 1D y 2D
 *
 * Esta cabecera es el contrato: el harness y tu implementacion
 * deben incluirla. Si cambias una firma, el harness dejara de compilar.
 */
#ifndef API_SESION27_H
#define API_SESION27_H

int emitir_acceso(char lineas[][64], int max, const char *base, const char *indice, int ancho, int *temporales);
int emitir_acceso_2d(char lineas[][64], int max, const char *base, const char *i, const char *j, int filas, int ancho, int *temporales);

#endif /* API_SESION27_H */
````

````python
%%writefile artesano_sesion27_c.c

#include "api_sesion27.h"
#include <stdio.h>
#include <limits.h>

int emitir_acceso(char lineas[][64], int max,
                  const char *base, const char *indice,
                  int ancho, int *temporales) {
    int t;
    int r;

    if (base == NULL) return 0;

    if (lineas == NULL || indice == NULL ||
        temporales == NULL || max < 2 ||
        ancho <= 0 || *temporales < 0 ||
        *temporales > INT_MAX - 2) {
        return 0;
    }

    t = *temporales;

    r = snprintf(lineas[0], 64, "_t%d = %s * %d",
                 t, indice, ancho);
    if (r < 0 || r >= 64) return 0;

    r = snprintf(lineas[1], 64, "_t%d = %s[_t%d]",
                 t + 1, base, t);
    if (r < 0 || r >= 64) return 0;

    *temporales += 2;
    return 2;
}

int emitir_acceso_2d(char lineas[][64], int max,
                     const char *base, const char *i,
                     const char *j, int filas, int ancho,
                     int *temporales) {
    int t;
    int r;

    if (base == NULL) return 0;

    if (lineas == NULL || i == NULL || j == NULL ||
        temporales == NULL || max < 4 ||
        filas <= 0 || ancho <= 0 ||
        *temporales < 0 || *temporales > INT_MAX - 4) {
        return 0;
    }

    t = *temporales;

    r = snprintf(lineas[0], 64, "_t%d = %s * %d",
                 t, i, filas);
    if (r < 0 || r >= 64) return 0;

    r = snprintf(lineas[1], 64, "_t%d = _t%d + %s",
                 t + 1, t, j);
    if (r < 0 || r >= 64) return 0;

    r = snprintf(lineas[2], 64, "_t%d = _t%d * %d",
                 t + 2, t + 1, ancho);
    if (r < 0 || r >= 64) return 0;

    r = snprintf(lineas[3], 64, "_t%d = %s[_t%d]",
                 t + 3, base, t + 2);
    if (r < 0 || r >= 64) return 0;

    *temporales += 4;
    return 4;
}
````

Requisitos para emitir_acceso:
1. Calcular el desplazamiento multiplicando el índice por el ancho.
2. Generar la instrucción de lectura base[offset].
3. Utilizar dos temporales consecutivos.
4. Incrementar el contador de temporales en dos.

Requisitos para emitir_acceso_2d:
1. Calcular i * filas.
2. Sumar j al resultado anterior.
3. Multiplicar el índice lineal por el ancho.
4. Generar la lectura base[offset].
5. Utilizar cuatro temporales consecutivos.
6. Incrementar el contador de temporales en cuatro.

Restricciones:
- Mantener exactamente las firmas originales.
- Si base es NULL, no generar instrucciones y devolver 0.
- Validar apuntadores, dimensiones, ancho y capacidad.
- Cada instrucción debe caber en 64 caracteres.
- Evitar desbordamiento del contador de temporales.
- Comprobar el resultado de snprintf().
- Evitar modificaciones parciales del arreglo lineas.
- No utilizar memoria dinámica.
- No incluir main().
- Compatible con C11 y GCC usando -Wall -Wextra -Werror.

**OUTPUT:** Entrega código modular, explicación del funcionamiento,
mejoras respecto a una versión manual y análisis de
complejidad temporal y espacial.

````python
%%writefile ia_sesion27_c.c

#include "api_sesion27.h"
#include <stdio.h>
#include <string.h>
#include <limits.h>

int emitir_acceso(char lineas[][64], int max,
                  const char *base, const char *indice,
                  int ancho, int *temporales) {
    char temp[2][64] = {{0}};
    int t;
    int r;

    if (base == NULL) return 0;

    if (lineas == NULL || indice == NULL ||
        temporales == NULL || max < 2 ||
        ancho <= 0 || *temporales < 0 ||
        *temporales > INT_MAX - 2) {
        return 0;
    }

    t = *temporales;

    r = snprintf(temp[0], sizeof(temp[0]),
                 "_t%d = %s * %d", t, indice, ancho);
    if (r < 0 || (size_t)r >= sizeof(temp[0])) {
        return 0;
    }

    r = snprintf(temp[1], sizeof(temp[1]),
                 "_t%d = %s[_t%d]", t + 1, base, t);
    if (r < 0 || (size_t)r >= sizeof(temp[1])) {
        return 0;
    }

    memcpy(lineas, temp, sizeof(temp));

    *temporales += 2;
    return 2;
}

int emitir_acceso_2d(char lineas[][64], int max,
                     const char *base, const char *i,
                     const char *j, int filas, int ancho,
                     int *temporales) {
    char temp[4][64] = {{0}};
    int t;
    int r;

    if (base == NULL) return 0;

    if (lineas == NULL || i == NULL || j == NULL ||
        temporales == NULL || max < 4 ||
        filas <= 0 || ancho <= 0 ||
        *temporales < 0 || *temporales > INT_MAX - 4) {
        return 0;
    }

    t = *temporales;

    /* Paso 1: calcular i * filas. */
    r = snprintf(temp[0], sizeof(temp[0]),
                 "_t%d = %s * %d", t, i, filas);
    if (r < 0 || (size_t)r >= sizeof(temp[0])) {
        return 0;
    }

    /* Paso 2: sumar j al resultado anterior. */
    r = snprintf(temp[1], sizeof(temp[1]),
                 "_t%d = _t%d + %s", t + 1, t, j);
    if (r < 0 || (size_t)r >= sizeof(temp[1])) {
        return 0;
    }

    /* Paso 3: multiplicar el índice lineal por el ancho. */
    r = snprintf(temp[2], sizeof(temp[2]),
                 "_t%d = _t%d * %d", t + 2, t + 1, ancho);
    if (r < 0 || (size_t)r >= sizeof(temp[2])) {
        return 0;
    }

    /* Paso 4: leer el elemento del arreglo. */
    r = snprintf(temp[3], sizeof(temp[3]),
                 "_t%d = %s[_t%d]", t + 3, base, t + 2);
    if (r < 0 || (size_t)r >= sizeof(temp[3])) {
        return 0;
    }

    memcpy(lineas, temp, sizeof(temp));

    *temporales += 4;
    return 4;
}
````

````python
%%writefile harness_sesion27_c.c
/* harness_sesion27_c.c - auditoria automatica de la Sesion 27
 *
 * Compilacion:
 *   gcc -std=c11 -Wall -Wextra -Werror harness_sesion27_c.c artesano_sesion27_c.c -o test_artesano
 *
 * El harness NO aprueba nada por si solo: cada linea se calcula
 * ejecutando tu implementacion y comparandola contra el valor esperado.
 * Si algun caso falla, el programa termina con codigo de salida 1.
 */
#include <stdio.h>
#include <string.h>
#include "api_sesion27.h"

static int probadas = 0;
static int fallidas = 0;

static void verificar_str(const char *caso, const char *esperado,
                          const char *obtenido) {
    int ok = (obtenido != NULL) && (strcmp(esperado, obtenido) == 0);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %-34s esperado: %-12s obtenido: %-12s %s\n",
           probadas, caso, esperado, obtenido ? obtenido : "(NULL)",
           ok ? "OK" : "FALLA");
}

static void verificar_str_largo(const char *caso, const char *esperado,
                                const char *obtenido) {
    int ok = (obtenido != NULL) && (strcmp(esperado, obtenido) == 0);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %s\n", probadas, caso);
    printf("     esperado: %s\n", esperado);
    printf("     obtenido: %s\n", obtenido ? obtenido : "(NULL)");
    printf("     -> %s\n", ok ? "OK" : "FALLA");
}

static void verificar_int(const char *caso, long esperado, long obtenido) {
    int ok = (esperado == obtenido);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %-34s esperado: %-12ld obtenido: %-12ld %s\n",
           probadas, caso, esperado, obtenido, ok ? "OK" : "FALLA");
}

int main(void) {
    printf("=== AUDITORIA SESION 27 - Desplazamientos para arreglos 1D y 2D ===\n");
    printf("Caso                               esperado:              obtenido:              resultado\n");

    { char l[8][64] = {{0}}; int t = 0; emitir_acceso(l, 8, "A", "i", 4, &t); verificar_str("1D: i * 4", "_t0 = i * 4", l[0]); }
    { char l[8][64] = {{0}}; int t = 0; emitir_acceso(l, 8, "A", "i", 4, &t); verificar_str("1D: A[_t0]", "_t1 = A[_t0]", l[1]); }
    { char l[16][64] = {{0}}; int t = 0; emitir_acceso_2d(l, 16, "M", "i", "j", 3, 4, &t); verificar_str("2D: i * 3", "_t0 = i * 3", l[0]); }
    { char l[16][64] = {{0}}; int t = 0; emitir_acceso_2d(l, 16, "M", "i", "j", 3, 4, &t); verificar_str_largo("2D: offset por ancho", "_t2 = _t1 * 4", l[2]); }
    { char l[16][64] = {{0}}; int t = 0; emitir_acceso_2d(l, 16, "M", "i", "j", 3, 4, &t); verificar_str("2D: M[_t2]", "_t3 = M[_t2]", l[3]); }
    { char l[8][64] = {{0}}; int t = 0; int k = emitir_acceso(l, 8, NULL, "i", 4, &t); verificar_int("base NULL (sin acceso)", (long)(0), (long)(k)); }

    printf("\nRESULTADO: %d/%d pruebas superadas\n",
           probadas - fallidas, probadas);
    if (fallidas != 0) {
        printf("Aun hay %d caso(s) sin resolver. Revisa tu implementacion.\n",
               fallidas);
        return 1;
    }
    printf("TODO EN VERDE. Documenta tu auditoria en auditorias/auditoria_sesion27.md" "\n");
    return 0;
}
````
