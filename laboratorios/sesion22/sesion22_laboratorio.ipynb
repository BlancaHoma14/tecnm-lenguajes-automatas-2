````python
%%writefile api_sesion22.h
/* api_sesion22.h - contrato de la Sesion 22
 * Traductor de AST a TAC con precedencia
 *
 * Esta cabecera es el contrato: el harness y tu implementacion
 * deben incluirla. Si cambias una firma, el harness dejara de compilar.
 */
#ifndef API_SESION22_H
#define API_SESION22_H

typedef struct Nodo {
    char tipo;
    char op;
    char valor[16];
    struct Nodo *izq;
    struct Nodo *der;
} Nodo;

int generar_tac(const Nodo *raiz, char lineas[][64], int max);

#endif /* API_SESION22_H */
````

````python
%%writefile artesano_sesion22_c.c

/* artesano_sesion22_c.c - Fase 1: El Artesano */

#include "api_sesion22.h"
#include <stdio.h>
#include <string.h>

typedef struct {
    char nombre[64][16];
    int usados;
} Pool;

/* Genera los temporales consecutivos */
const char *nuevo_temporal(Pool *p, int *contador) {
    if (p->usados >= 64) {
        return "_t?";
    }

    snprintf(p->nombre[p->usados],
             sizeof p->nombre[0],
             "_t%d", (*contador)++);

    return p->nombre[p->usados++];
}

/* Estado para controlar la traducción */
typedef struct {
    char (*lineas)[64];
    int max;
    int cantidad;
    int contador;
    Pool pool;
} Estado;

/* Traduce el AST utilizando postorden */
static const char *traducir(const Nodo *n, Estado *e) {

    const char *izq;
    const char *der;
    const char *temp;

    if (n == NULL) {
        return NULL;
    }

    /* Si es una hoja, devuelve su valor */
    if (n->tipo == 'H') {
        return n->valor;
    }

    /* Si no es un operador binario, no se procesa */
    if (n->tipo != 'B') {
        return NULL;
    }

    /* Primero se procesa el hijo izquierdo */
    izq = traducir(n->izq, e);

    if (izq == NULL) {
        return NULL;
    }

    /* Después se procesa el hijo derecho */
    der = traducir(n->der, e);

    if (der == NULL) {
        return NULL;
    }

    /* Verifica que todavía haya espacio */
    if (e->cantidad >= e->max ||
        e->pool.usados >= 64) {
        return NULL;
    }

    /* Genera un temporal */
    temp = nuevo_temporal(&e->pool, &e->contador);

    /* Escribe la instrucción TAC */
    snprintf(e->lineas[e->cantidad], 64,
             "%s = %s %c %s",
             temp, izq, n->op, der);

    e->cantidad++;

    return temp;
}

/* Función solicitada por el contrato */
int generar_tac(const Nodo *raiz, char lineas[][64], int max) {

    Estado estado;

    if (raiz == NULL || lineas == NULL || max <= 0) {
        return 0;
    }

    estado.lineas = lineas;
    estado.max = max;
    estado.cantidad = 0;
    estado.contador = 0;
    estado.pool.usados = 0;

    traducir(raiz, &estado);

    return estado.cantidad;
}
````

````python
%%writefile ia_sesion22_c.c

/* ia_sesion22_c.c - Fase 2: El Cientifico */

#include "api_sesion22.h"
#include <stdio.h>
#include <string.h>

#define MAX_TEMPORALES 64
#define TAM_TEMPORAL 16
#define TAM_LINEA 64

typedef struct {
    char nombre[MAX_TEMPORALES][TAM_TEMPORAL];
    int usados;
} Pool;

typedef struct {
    char (*lineas)[TAM_LINEA];
    int max;
    int cantidad;
    int contador;
    Pool pool;
    int error;
} Estado;

/* Genera un nuevo temporal */
static const char *nuevo_temporal(Pool *p, int *contador) {

    if (p == NULL || contador == NULL ||
        p->usados >= MAX_TEMPORALES) {
        return NULL;
    }

    int escrito = snprintf(
        p->nombre[p->usados],
        sizeof p->nombre[0],
        "_t%d",
        *contador
    );

    if (escrito < 0 ||
        escrito >= (int)sizeof p->nombre[0]) {
        return NULL;
    }

    (*contador)++;

    return p->nombre[p->usados++];
}

/* Traducción recursiva del AST */
static const char *traducir(const Nodo *n, Estado *e) {

    const char *izq;
    const char *der;
    const char *temp;
    int escrito;

    if (n == NULL || e == NULL || e->error) {
        if (e != NULL) {
            e->error = 1;
        }
        return NULL;
    }

    /* Procesamiento de hojas */
    if (n->tipo == 'H') {

        if (n->valor[0] == '\0') {
            e->error = 1;
            return NULL;
        }

        return n->valor;
    }

    /* Validación del tipo de nodo */
    if (n->tipo != 'B' || n->op == '\0') {
        e->error = 1;
        return NULL;
    }

    /* Recorrido en postorden: hijo izquierdo */
    izq = traducir(n->izq, e);

    if (izq == NULL || e->error) {
        return NULL;
    }

    /* Recorrido en postorden: hijo derecho */
    der = traducir(n->der, e);

    if (der == NULL || e->error) {
        return NULL;
    }

    /* Verifica la capacidad antes de emitir */
    if (e->cantidad >= e->max ||
        e->pool.usados >= MAX_TEMPORALES) {
        e->error = 1;
        return NULL;
    }

    /* Solicita un temporal */
    temp = nuevo_temporal(&e->pool, &e->contador);

    if (temp == NULL) {
        e->error = 1;
        return NULL;
    }

    /* Emite la instrucción TAC */
    escrito = snprintf(
        e->lineas[e->cantidad],
        TAM_LINEA,
        "%s = %s %c %s",
        temp, izq, n->op, der
    );

    if (escrito < 0 || escrito >= TAM_LINEA) {
        e->lineas[e->cantidad][0] = '\0';
        e->error = 1;
        return NULL;
    }

    e->cantidad++;

    return temp;
}

/* Interfaz pública */
int generar_tac(const Nodo *raiz, char lineas[][64], int max) {

    Estado estado;

    if (raiz == NULL || lineas == NULL || max <= 0) {
        return 0;
    }

    estado.lineas = lineas;
    estado.max = max;
    estado.cantidad = 0;
    estado.contador = 0;
    estado.pool.usados = 0;
    estado.error = 0;

    traducir(raiz, &estado);

    return estado.cantidad;
}
````
