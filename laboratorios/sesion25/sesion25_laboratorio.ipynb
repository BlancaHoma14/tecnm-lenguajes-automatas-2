````python
%%writefile api_sesion25.h
/* api_sesion25.h - contrato de la Sesion 25
 * Esqueleto TAC de while y for
 *
 * Esta cabecera es el contrato: el harness y tu implementacion
 * deben incluirla. Si cambias una firma, el harness dejara de compilar.
 */
#ifndef API_SESION25_H
#define API_SESION25_H

void emitir_while(char lineas[][64], int max, int *n, const char *cond, int *etiquetas);
void emitir_for(char lineas[][64], int max, int *n, const char *inicio, const char *cond, const char *post, int *etiquetas);

#endif /* API_SESION25_H */
````

````python
%%writefile artesano_sesion25_c.c
#include "api_sesion25.h"
#include <stdio.h>
#include <limits.h>

void emitir_while(char lineas[][64], int max,
                  int *n, const char *cond, int *etiquetas) {
    int inicio;
    int fin;
    int r;

    if (lineas == NULL || n == NULL || cond == NULL ||
        etiquetas == NULL || max < 0 ||
        *n < 0 || *n > max ||
        *etiquetas < 0 || *etiquetas > INT_MAX - 2) {
        return;
    }

    if (max - *n < 4) {
        return;
    }

    inicio = *etiquetas;
    fin = inicio + 1;

    r = snprintf(lineas[*n], 64, "L%d:", inicio);
    if (r < 0 || r >= 64) return;
    (*n)++;

    r = snprintf(lineas[*n], 64,
                 "IFFALSE %s GOTO L%d", cond, fin);
    if (r < 0 || r >= 64) {
        (*n)--;
        lineas[*n][0] = '\0';
        return;
    }
    (*n)++;

    r = snprintf(lineas[*n], 64, "GOTO L%d", inicio);
    if (r < 0 || r >= 64) {
        (*n)--;
        lineas[*n][0] = '\0';
        (*n)--;
        lineas[*n][0] = '\0';
        return;
    }
    (*n)++;

    r = snprintf(lineas[*n], 64, "L%d:", fin);
    if (r < 0 || r >= 64) {
        (*n)--;
        lineas[*n][0] = '\0';
        (*n)--;
        lineas[*n][0] = '\0';
        (*n)--;
        lineas[*n][0] = '\0';
        return;
    }
    (*n)++;

    *etiquetas += 2;
}

void emitir_for(char lineas[][64], int max, int *n,
                const char *inicio, const char *cond,
                const char *post, int *etiquetas) {
    int etiqueta_inicio;
    int etiqueta_fin;
    int cantidad;
    int r;

    if (lineas == NULL || n == NULL || cond == NULL ||
        etiquetas == NULL || max < 0 ||
        *n < 0 || *n > max ||
        *etiquetas < 0 || *etiquetas > INT_MAX - 2) {
        return;
    }

    cantidad = 4;
    if (inicio != NULL) cantidad++;
    if (post != NULL) cantidad++;

    if (max - *n < cantidad) {
        return;
    }

    etiqueta_inicio = *etiquetas;
    etiqueta_fin = etiqueta_inicio + 1;

    if (inicio != NULL) {
        r = snprintf(lineas[*n], 64, "%s", inicio);
        if (r < 0 || r >= 64) return;
        (*n)++;
    }

    r = snprintf(lineas[*n], 64, "L%d:", etiqueta_inicio);
    if (r < 0 || r >= 64) return;
    (*n)++;

    r = snprintf(lineas[*n], 64,
                 "IFFALSE %s GOTO L%d", cond, etiqueta_fin);
    if (r < 0 || r >= 64) return;
    (*n)++;

    if (post != NULL) {
        r = snprintf(lineas[*n], 64, "%s", post);
        if (r < 0 || r >= 64) return;
        (*n)++;
    }

    r = snprintf(lineas[*n], 64, "GOTO L%d", etiqueta_inicio);
    if (r < 0 || r >= 64) return;
    (*n)++;

    r = snprintf(lineas[*n], 64, "L%d:", etiqueta_fin);
    if (r < 0 || r >= 64) return;
    (*n)++;

    *etiquetas += 2;
}
````

Requisitos para emitir_while():

Restricciones generales:

````python
%%writefile ia_sesion25_c.c

#include "api_sesion25.h"
#include <stdio.h>
#include <string.h>
#include <limits.h>

void emitir_while(char lineas[][64], int max,
                  int *n, const char *cond, int *etiquetas) {
    char temp[4][64] = {{0}};
    int inicio;
    int fin;
    int r;

    /* Validar los argumentos */
    if (lineas == NULL || n == NULL || cond == NULL ||
        etiquetas == NULL || max < 0 ||
        *n < 0 || *n > max ||
        *etiquetas < 0 || *etiquetas > INT_MAX - 2) {
        return;
    }

    /* El while necesita cuatro líneas */
    if (max - *n < 4) {
        return;
    }

    inicio = *etiquetas;
    fin = inicio + 1;

    r = snprintf(temp[0], sizeof(temp[0]), "L%d:", inicio);
    if (r < 0 || (size_t)r >= sizeof(temp[0])) return;

    r = snprintf(temp[1], sizeof(temp[1]),
                 "IFFALSE %s GOTO L%d", cond, fin);
    if (r < 0 || (size_t)r >= sizeof(temp[1])) return;

    r = snprintf(temp[2], sizeof(temp[2]), "GOTO L%d", inicio);
    if (r < 0 || (size_t)r >= sizeof(temp[2])) return;

    r = snprintf(temp[3], sizeof(temp[3]), "L%d:", fin);
    if (r < 0 || (size_t)r >= sizeof(temp[3])) return;

    /* Copiar solo después de generar todo correctamente */
    memcpy(&lineas[*n], temp, sizeof(temp));

    *n += 4;
    *etiquetas += 2;
}

void emitir_for(char lineas[][64], int max, int *n,
                const char *inicio, const char *cond,
                const char *post, int *etiquetas) {
    char temp[6][64] = {{0}};
    int etiqueta_inicio;
    int etiqueta_fin;
    int cantidad = 0;
    int r;

    /* Validar los argumentos */
    if (lineas == NULL || n == NULL || cond == NULL ||
        etiquetas == NULL || max < 0 ||
        *n < 0 || *n > max ||
        *etiquetas < 0 || *etiquetas > INT_MAX - 2) {
        return;
    }

    etiqueta_inicio = *etiquetas;
    etiqueta_fin = etiqueta_inicio + 1;

    /* Inicialización opcional */
    if (inicio != NULL) {
        r = snprintf(temp[cantidad], sizeof(temp[cantidad]),
                     "%s", inicio);
        if (r < 0 || (size_t)r >= sizeof(temp[cantidad])) return;
        cantidad++;
    }

    /* Etiqueta de inicio */
    r = snprintf(temp[cantidad], sizeof(temp[cantidad]),
                 "L%d:", etiqueta_inicio);
    if (r < 0 || (size_t)r >= sizeof(temp[cantidad])) return;
    cantidad++;

    /* Condición de salida */
    r = snprintf(temp[cantidad], sizeof(temp[cantidad]),
                 "IFFALSE %s GOTO L%d", cond, etiqueta_fin);
    if (r < 0 || (size_t)r >= sizeof(temp[cantidad])) return;
    cantidad++;

    /* Actualización opcional */
    if (post != NULL) {
        r = snprintf(temp[cantidad], sizeof(temp[cantidad]),
                     "%s", post);
        if (r < 0 || (size_t)r >= sizeof(temp[cantidad])) return;
        cantidad++;
    }

    /* Salto de regreso */
    r = snprintf(temp[cantidad], sizeof(temp[cantidad]),
                 "GOTO L%d", etiqueta_inicio);
    if (r < 0 || (size_t)r >= sizeof(temp[cantidad])) return;
    cantidad++;

    /* Etiqueta de salida */
    r = snprintf(temp[cantidad], sizeof(temp[cantidad]),
                 "L%d:", etiqueta_fin);
    if (r < 0 || (size_t)r >= sizeof(temp[cantidad])) return;
    cantidad++;

    /* Verificar capacidad antes de modificar el arreglo final */
    if (max - *n < cantidad) {
        return;
    }

    /* Copiar las instrucciones generadas */
    memcpy(&lineas[*n], temp,
           (size_t)cantidad * sizeof(temp[0]));

    *n += cantidad;
    *etiquetas += 2;
}
````

````python
%%writefile harness_sesion25_c.c
/* harness_sesion25_c.c - auditoria automatica de la Sesion 25
 *
 * Compilacion:
 *   gcc -std=c11 -Wall -Wextra -Werror harness_sesion25_c.c artesano_sesion25_c.c -o test_artesano
 *
 * El harness NO aprueba nada por si solo: cada linea se calcula
 * ejecutando tu implementacion y comparandola contra el valor esperado.
 * Si algun caso falla, el programa termina con codigo de salida 1.
 */
#include <stdio.h>
#include <string.h>
#include "api_sesion25.h"

static int probadas = 0;
static int fallidas = 0;

static void verificar_str(const char *caso, const char *esperado,
                          const char *obtenido) {
    int ok = (obtenido != NULL) && (strcmp(esperado, obtenido) == 0);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %-34s esperado: %-12s obtenido: %-12s %s\n",
           probadas, caso, esperado, obtenido ? obtenido : "(NULL)",
           ok ? "OK" : "FALLA");
}

static void verificar_str_largo(const char *caso, const char *esperado,
                                const char *obtenido) {
    int ok = (obtenido != NULL) && (strcmp(esperado, obtenido) == 0);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %s\n", probadas, caso);
    printf("     esperado: %s\n", esperado);
    printf("     obtenido: %s\n", obtenido ? obtenido : "(NULL)");
    printf("     -> %s\n", ok ? "OK" : "FALLA");
}

static void verificar_int(const char *caso, long esperado, long obtenido) {
    int ok = (esperado == obtenido);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %-34s esperado: %-12ld obtenido: %-12ld %s\n",
           probadas, caso, esperado, obtenido, ok ? "OK" : "FALLA");
}

int main(void) {
    printf("=== AUDITORIA SESION 25 - Esqueleto TAC de while y for ===\n");
    printf("Caso                               esperado:              obtenido:              resultado\n");

    { char l[16][64] = {{0}}; int n = 0, e = 0; emitir_while(l, 16, &n, "a < b", &e); verificar_str("while: marca de inicio", "L0:", l[0]); }
    { char l[16][64] = {{0}}; int n = 0, e = 0; emitir_while(l, 16, &n, "a < b", &e); verificar_str_largo("while: condicion", "IFFALSE a < b GOTO L1", l[1]); }
    { char l[16][64] = {{0}}; int n = 0, e = 0; emitir_while(l, 16, &n, "a < b", &e); emitir_while(l, 16, &n, "c < d", &e); verificar_str("dos ciclos sin colision", "L2:", l[4]); }
    { char l[16][64] = {{0}}; int n = 0, e = 0; emitir_for(l, 16, &n, "i = 0", "i < 10", "i = i + 1", &e); verificar_str("for: inicializacion", "i = 0", l[0]); }
    { char l[16][64] = {{0}}; int n = 0, e = 0; emitir_for(l, 16, &n, "i = 0", "i < 10", "i = i + 1", &e); verificar_str("for: actualizacion", "i = i + 1", l[3]); }
    { char l[16][64] = {{0}}; int n = 0, e = 0; emitir_for(l, 16, &n, "j = 0", "j < 3", NULL, &e); verificar_int("for sin post (NULL)", (long)(5), (long)(n)); }

    printf("\nRESULTADO: %d/%d pruebas superadas\n",
           probadas - fallidas, probadas);
    if (fallidas != 0) {
        printf("Aun hay %d caso(s) sin resolver. Revisa tu implementacion.\n",
               fallidas);
        return 1;
    }
    printf("TODO EN VERDE. Documenta tu auditoria en auditorias/auditoria_sesion25.md" "\n");
    return 0;
}
````
