- [ ] Un árbol nulo produce cadena vacía y el buffer siempre termina en '\0'.

## Fase 1 · El Artesano (15 min, sin IA)

Escribe a mano la implementación en `artesano_sesion05_c.c`. El archivo ya trae la

````python
%%writefile api_sesion05.h
/* api_sesion05.h - contrato de la Sesion 05
 * Recorrido postorden y RPN
 *
 * Esta cabecera es el contrato: el harness y tu implementacion
 * deben incluirla. Si cambias una firma, el harness dejara de compilar.
 */
#ifndef API_SESION05_H
#define API_SESION05_H

typedef enum { NODO_NUM, NODO_BINOP } TipoNodo;
typedef struct ASTNode {
    TipoNodo tipo;
    int valor;
    char op;
    struct ASTNode *izq;
    struct ASTNode *der;
} ASTNode;

void postorden(ASTNode *n, char *buf, int cap);

#endif /* API_SESION05_H */
````

````python
%%writefile artesano_sesion05_c.c
/* artesano_sesion05_c.c - Fase 1: El Artesano */
#include "api_sesion05.h"
#include <stdio.h>
#include <string.h>

void postorden(ASTNode *n, char *buf, int cap) {
    if (buf == NULL || cap <= 0) {
        return;
    }

    buf[0] = '\0';

    if (n == NULL) {
        return;
    }

    if (n->tipo == NODO_NUM) {
        snprintf(buf, (size_t)cap, "%d", n->valor);
        return;
    }

    if (n->tipo == NODO_BINOP) {
        char izq[256] = "";
        char der[256] = "";
        char raiz[2] = {n->op, '\0'};

        postorden(n->izq, izq, (int)sizeof(izq));
        postorden(n->der, der, (int)sizeof(der));

        if (izq[0] != '\0' && der[0] != '\0') {
            snprintf(buf, (size_t)cap, "%s %s %s",
                     izq, der, raiz);
        } else if (izq[0] != '\0') {
            snprintf(buf, (size_t)cap, "%s %s", izq, raiz);
        } else if (der[0] != '\0') {
            snprintf(buf, (size_t)cap, "%s %s", der, raiz);
        } else {
            snprintf(buf, (size_t)cap, "%s", raiz);
        }
    }
}
````

**C — Contexto**: Estoy trabajando en Lenguajes y Autómatas II. Debo recorrer un árbol sintáctico abstracto (AST) en postorden y generar su representación en notación polaca inversa (RPN).

**E — Especificación**: Implementa postorden(ASTNode *n, char *buf, int cap) respetando exactamente la declaración de api_sesion05.h. Recorre primero el hijo izquierdo, después el derecho y al final la raíz. Los números y operadores deben separarse con un solo espacio, sin espacio final. Si el árbol es nulo, produce una cadena vacía. Respeta la capacidad del buffer, garantiza el terminador '\0' cuando buf y cap lo permitan, y evita errores y advertencias con -Wall -Wextra -Werror. No agregues main().

**O — Output**: Entrega el código completo de ia_sesion05_c.c, listo para compilar con el harness de la Sesión 05, y explica las mejoras respecto a mi implementación manual.

````python
%%writefile ia_sesion05_c.c
/* ia_sesion05_c.c - Fase 2: El Cientifico */
#include "api_sesion05.h"
#include <stdio.h>
#include <string.h>

/* Estado del recorrido y control de capacidad. */
typedef struct {
    char *buf;
    int cap;
    int pos;
    int error;
} EstadoRPN;

/* Agrega un token completo, sin exceder la capacidad. */
static void agregar_token(EstadoRPN *e, const char *token) {
    size_t longitud;
    size_t separador;
    size_t disponible;

    if (e == NULL || token == NULL || e->error) {
        return;
    }

    longitud = strlen(token);
    separador = (e->pos > 0) ? 1U : 0U;
    disponible = (size_t)(e->cap - e->pos - 1);

    if (longitud + separador > disponible) {
        e->error = 1;
        return;
    }

    if (separador != 0U) {
        e->buf[e->pos++] = ' ';
    }

    memcpy(e->buf + e->pos, token, longitud);
    e->pos += (int)longitud;
    e->buf[e->pos] = '\0';
}

/* Recorre el AST en postorden: izquierdo, derecho, raíz. */
static void recorrer(ASTNode *n, EstadoRPN *e) {
    char token[32];

    if (n == NULL || e == NULL || e->error) {
        return;
    }

    if (n->tipo == NODO_NUM) {
        (void)snprintf(token, sizeof(token), "%d", n->valor);
        agregar_token(e, token);
        return;
    }

    if (n->tipo == NODO_BINOP) {
        recorrer(n->izq, e);
        recorrer(n->der, e);

        if (e->error) {
            return;
        }

        token[0] = n->op;
        token[1] = '\0';
        agregar_token(e, token);
    }
}

void postorden(ASTNode *n, char *buf, int cap) {
    EstadoRPN estado;

    if (buf == NULL || cap <= 0) {
        return;
    }

    buf[0] = '\0';

    estado.buf = buf;
    estado.cap = cap;
    estado.pos = 0;
    estado.error = 0;

    recorrer(n, &estado);
}
````

````python
%%writefile harness_sesion05_c.c
/* harness_sesion05_c.c - auditoria automatica de la Sesion 05
 *
 * Compilacion:
 *   gcc -std=c11 -Wall -Wextra -Werror harness_sesion05_c.c artesano_sesion05_c.c -o test_artesano
 *
 * El harness NO aprueba nada por si solo: cada linea se calcula
 * ejecutando tu implementacion y comparandola contra el valor esperado.
 * Si algun caso falla, el programa termina con codigo de salida 1.
 */
#include <stdio.h>
#include <string.h>
#include "api_sesion05.h"

static int probadas = 0;
static int fallidas = 0;

static void verificar_str(const char *caso, const char *esperado,
                          const char *obtenido) {
    int ok = (obtenido != NULL) && (strcmp(esperado, obtenido) == 0);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %-34s esperado: %-12s obtenido: %-12s %s\n",
           probadas, caso, esperado, obtenido ? obtenido : "(NULL)",
           ok ? "OK" : "FALLA");
}

static void verificar_str_largo(const char *caso, const char *esperado,
                                const char *obtenido) {
    int ok = (obtenido != NULL) && (strcmp(esperado, obtenido) == 0);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %s\n", probadas, caso);
    printf("     esperado: %s\n", esperado);
    printf("     obtenido: %s\n", obtenido ? obtenido : "(NULL)");
    printf("     -> %s\n", ok ? "OK" : "FALLA");
}

int main(void) {
    printf("=== AUDITORIA SESION 05 - Recorrido postorden y RPN ===\n");
    printf("Caso                               esperado:              obtenido:              resultado\n");

    { ASTNode a = {NODO_NUM, 42, 0, NULL, NULL}; char b[64]; postorden(&a, b, 64); verificar_str("hoja 42", "42", b); }
    { ASTNode a = {NODO_NUM, 3, 0, NULL, NULL}; ASTNode b = {NODO_NUM, 5, 0, NULL, NULL}; ASTNode c = {NODO_NUM, 2, 0, NULL, NULL}; ASTNode m = {NODO_BINOP, 0, '*', &b, &c}; ASTNode s = {NODO_BINOP, 0, '+', &a, &m}; char buf[64]; postorden(&s, buf, 64); verificar_str("3+(5*2)", "3 5 2 * +", buf); }
    { char b[16]; postorden(NULL, b, 16); verificar_str("arbol NULL", "", b); }
    { ASTNode a = {NODO_NUM, 1, 0, NULL, NULL}; ASTNode b = {NODO_NUM, 2, 0, NULL, NULL}; ASTNode c = {NODO_NUM, 3, 0, NULL, NULL}; ASTNode d = {NODO_NUM, 4, 0, NULL, NULL}; ASTNode s1 = {NODO_BINOP, 0, '+', &a, &b}; ASTNode s2 = {NODO_BINOP, 0, '+', &c, &d}; ASTNode m = {NODO_BINOP, 0, '*', &s1, &s2}; char buf[64]; postorden(&m, buf, 64); verificar_str_largo("(1+2)*(3+4)", "1 2 + 3 4 + *", buf); }
    { ASTNode a = {NODO_NUM, 8, 0, NULL, NULL}; ASTNode b = {NODO_NUM, 3, 0, NULL, NULL}; ASTNode r = {NODO_BINOP, 0, '-', &a, &b}; char buf[32]; postorden(&r, buf, 32); verificar_str("8-3", "8 3 -", buf); }

    printf("\nRESULTADO: %d/%d pruebas superadas\n",
           probadas - fallidas, probadas);
    if (fallidas != 0) {
        printf("Aun hay %d caso(s) sin resolver. Revisa tu implementacion.\n",
               fallidas);
        return 1;
    }
    printf("TODO EN VERDE. Documenta tu auditoria en auditorias/auditoria_sesion05.md" "\n");
    return 0;
}
````
